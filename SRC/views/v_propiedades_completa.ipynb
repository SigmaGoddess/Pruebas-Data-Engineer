```sql
CREATE OR REPLACE VIEW bootcamp.semantica.v_propiedades_completa AS
SELECT 
    dz.partido,
    dz.region,
    dto.tipo_operacion,
    dto.moneda,
    dc.estado,
    dc.cochera,
    do.orientacion,
    fp.precio,
    fp.precio_por_m2,
    fp.metros_cuadrados_totales,
    fp.metros_cuadrados_cubiertos,
    fp.expensas,
    fp.ambientes,
    fp.url
FROM bootcamp.gold.fact_propiedades fp
JOIN bootcamp.gold.dim_zona dz ON fp.zona_id = dz.zona_id
JOIN bootcamp.gold.dim_tipo_operacion dto ON fp.tipo_operacion_id = dto.tipo_operacion_id
JOIN bootcamp.gold.dim_caracteristicas dc ON fp.caracteristicas_id = dc.caracteristicas_id
JOIN bootcamp.gold.dim_orientacion do ON fp.orientacion_id = do.orientacion_id;
```

```sql
SELECT * FROM bootcamp.semantica.v_propiedades_completa;
```
