```sql
CREATE OR REPLACE VIEW bootcamp.semantica.v_metricas_zona AS
SELECT
    dz.partido,
    dz.region,
    dto.tipo_operacion,
    COUNT(*) AS total_propiedades,
    ROUND(AVG(fp.precio), 2) AS precio_promedio,
    ROUND(MEDIAN(fp.precio), 2) AS precio_mediana,
    ROUND(MAX(fp.precio), 2) AS precio_maximo,
    ROUND(MIN(fp.precio), 2) AS precio_minimo,
    ROUND(AVG(fp.precio_por_m2), 2) AS precio_m2_promedio,
    ROUND(MEDIAN(fp.precio_por_m2), 2) AS precio_m2_mediana,
    ROUND(AVG(fp.metros_cuadrados_totales), 2) AS metros_cuadrados_promedio,
    ROUND(MEDIAN(fp.metros_cuadrados_totales), 2) AS metros_cuadrados_mediana,
    ROUND(AVG(fp.ambientes), 2) AS ambientes_promedio
FROM bootcamp.gold.fact_propiedades fp
JOIN bootcamp.gold.dim_zona dz ON fp.zona_id = dz.zona_id
JOIN bootcamp.gold.dim_tipo_operacion dto ON fp.tipo_operacion_id = dto.tipo_operacion_id
GROUP BY dz.partido, dz.region, dto.tipo_operacion;
```

```sql
SELECT * FROM bootcamp.semantica.v_metricas_zona;
```
