```sql
%sql
-- ETL desde fechas de _processing_timestamp
INSERT OVERWRITE bootcamp.gold.dim_tiempo (fecha_id, fecha, anio, mes, trimestre, dia_semana, es_fin_de_semana)
  SELECT DISTINCT
    CAST(DATE_FORMAT(fecha_publicacion, 'yyyyMMdd') AS BIGINT) AS fecha_id,
    fecha_publicacion AS fecha,
    YEAR(fecha_publicacion) AS anio,
    MONTH(fecha_publicacion) AS mes,
    QUARTER(fecha_publicacion) AS trimestre,
    DATE_FORMAT(fecha_publicacion, 'EEEE') AS dia_semana,
    CASE WHEN DAYOFWEEK(fecha_publicacion) IN (1, 7) THEN TRUE ELSE FALSE END AS es_fin_de_semana
  FROM bootcamp.silver.propiedades
  WHERE fecha_publicacion IS NOT NULL;
```

```sql
-- Verificar
SELECT COUNT(*) AS total_fechas, MIN(fecha) AS fecha_min, MAX(fecha) AS fecha_max
FROM bootcamp.gold.dim_tiempo;
```
