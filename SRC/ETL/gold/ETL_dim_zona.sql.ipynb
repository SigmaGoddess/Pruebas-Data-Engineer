```sql
-- ETL
INSERT OVERWRITE bootcamp.gold.dim_zona (partido, region, ciudad, provincia, pais)
    SELECT DISTINCT 
        sp.partido, 
        sp.region,
        CASE 
            WHEN region = 'capital federal' THEN 'CABA' 
            ELSE 'GBA' 
        END AS ciudad,
        'Buenos Aires' as provincia,
        'Argentina' as pais
    FROM bootcamp.silver.propiedades sp
    WHERE partido IS NOT NULL;
```
