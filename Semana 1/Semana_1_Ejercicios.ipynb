```sql
SELECT 
    MAX(trip_distance) AS mayor_distancia,
    MIN(trip_distance) AS menor_distancia
FROM samples.nyctaxi.trips
WHERE trip_distance > 0;
```

```sql
SELECT 
    ROUND(AVG(trip_distance), 2) AS promedio_distancia,
    ROUND(AVG(fare_amount), 2) AS promedio_tarifa
FROm samples.nyctaxi.trips;
```

```sql
-- Viajes recientes
SELECT 
    trip_distance as distancia, 
    fare_amount as valor,
    tpep_pickup_datetime as fecha_viaje
FROM samples.nyctaxi.trips
ORDER BY tpep_pickup_datetime DESC
LIMIT 20;
```

```sql
-- Múltiples condiciones
SELECT 
    trip_distance,
    fare_amount
FROM samples.nyctaxi.trips
WHERE trip_distance > 10
AND fare_amount > 10;
```

```sql
-- Condiciones con OR
SELECT 
    trip_distance,
    fare_amount
FROM samples.nyctaxi.trips
WHERE trip_distance > 20
OR fare_amount > 100;
```

```sql
-- Rango de valores
SELECT 
    trip_distance,
    fare_amount
FROM samples.nyctaxi.trips
WHERE trip_distance BETWEEN 10 AND 20;
```

```sql
-- Ejercicio 3.1: Conteo básico
SELECT 
    COUNT(*) AS total_viajes
FROM samples.nyctaxi.trips;
```

```sql
-- Ejercicio 3.2: Sumas
SELECT 
    ROUND(SUM(trip_distance), 2) AS total_distancia,
    ROUND(SUM(fare_amount), 2) AS total_tarifa
FROM samples.nyctaxi.trips;
```

```sql
-- Ejercicio 3.3: Promedios
SELECT 
    ROUND(AVG(trip_distance), 2) AS promedio_distancia,
    ROUND(AVG(fare_amount), 2) AS promedio_tarifa
FROM samples.nyctaxi.trips
WHERE trip_distance > 0;
```

```sql
-- Ejercicio 3.4: Mínimos y máximos
SELECT 
    MAX(trip_distance) AS max_distancia,
    MIN(trip_distance) AS min_distancia,
    MAX(fare_amount) AS max_tarifa,
    MIN(fare_amount) AS min_tarifa
FROM samples.nyctaxi.trips
WHERE trip_distance > 0;
```

```sql
-- Ejercicio 4.1: Agrupar por zona
SELECT 
SUM(fare_amount) as total_tarifa,
AVG(fare_amount) as tarifa_promedio,
pickup_zip,
COUNT(*) as total_viaje
FROM samples.nyctaxi.trips
GROUP BY pickup_zip
ORDER BY total_tarifa DESC
LIMIT 10;
```

```sql
-- Ejercicio 4.2: Viajes por día de la semana
SELECT
    DAYOFWEEK(tpep_pickup_datetime) AS dia_semana,
    COUNT(*) as total_viajes,
    ROUND(AVG(trip_distance), 2) AS promedio_distancia
FROM samples.nyctaxi.trips
GROUP BY dia_semana;
```

```sql
-- Ejercicio 4.3: Viajes por hora
SELECT
    HOUR(tpep_pickup_datetime) AS hora_dia,
    COUNT(*) as total_viajes,
    ROUND(AVG(trip_distance), 2) AS promedio_distancia
FROM samples.nyctaxi.trips
GROUP BY hora_dia
ORDER BY hora_dia;
```

```sql
-- Ejercicio 5.1: Filtrar grupos grandes
SELECT 
    pickup_zip,
    COUNT(*) AS total_viajes
FROM samples.nyctaxi.trips
GROUP BY pickup_zip
HAVING total_viajes > 1000
ORDER BY total_viajes DESC;
```

```sql
-- Ejercicio 5.2: Días con muchos viajes
SELECT 
    COUNT(*) AS total_viajes,
    ROUND(AVG(fare_amount), 2) AS tarifa_promedio,
    DAYOFWEEK(tpep_pickup_datetime) as dia_semana
FROM samples.nyctaxi.trips
GROUP BY dia_semana
HAVING total_viajes > 200
ORDER BY total_viajes DESC
LIMIT 10;
```

```sql
-- Ejercicio 5.3: Horas pico
SELECT 
    COUNT(*) AS total_viajes,
    hour(tpep_pickup_datetime) as hora_dia
FROM samples.nyctaxi.trips
GROUP BY hora_dia
HAVING total_viajes > 1000
ORDER BY total_viajes DESC;
```

```sql
-- Ejercicio 6.1: Análisis completo por zona
SELECT
    pickup_zip,
    COUNT(*) AS total_viajes,
    ROUND(MAX(fare_amount), 2) AS max_tarifa,
    ROUND(MIN(fare_amount), 2) AS min_tarifa,
    ROUND(AVG(fare_amount), 2) AS tarifa_promedio
FROM samples.nyctaxi.trips
WHERE fare_amount > 5 AND trip_distance > 0
GROUP BY pickup_zip
HAVING COUNT(*) > 1000
ORDER BY total_viajes DESC
LIMIT 10;
```

```sql
-- Ejercicio 6.3: Top viajes
SELECT
  tpep_pickup_datetime,
  trip_distance,
  fare_amount
FROM samples.nyctaxi.trips
WHERE trip_distance > 0
ORDER BY trip_distance DESC
LIMIT 10;
```

```sql
-- Ejercicio 6.2: Categorizar viajes por distancia
SELECT 
 CASE
    WHEN trip_distance < 2 THEN 'Corto (< 2 mi)'
    WHEN trip_distance BETWEEN 2 AND 5 THEN 'Medio (2-5 mi)'
    WHEN trip_distance BETWEEN 5 AND 10 THEN 'Largo (5-10 mi)'
    ELSE 'Muy largo (> 10 mi)'
  END as categoria,
  COUNT(*) as total_viajes,
  ROUND(AVG(fare_amount), 2) AS tarifa_promedio,
  ROUND(AVG(trip_distance), 2) AS promedio_distancia
FROM samples.nyctaxi.trips
GROUP BY categoria
HAVING COUNT(*) > 100;
```
